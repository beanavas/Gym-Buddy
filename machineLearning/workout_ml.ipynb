```python
def map_to_region(title):
    for exercise in exerciseLibrary:
        if exercise["name"].lower() == title.lower():
            return exercise["group"]
    return "Other"  # fallback if not found

# Apply to your DataFrame
df["region"] = df["title"].apply(map_to_region)

# Preview the result
df[["title", "region"]].head()
```

```python
all_regions = ["Legs", "Core", "Back", "Chest", "Shoulders", "Arms", "Glutes", "Full Body"]

# Equal weighting for each region (you can tweak later)
ideal_distribution = pd.Series(
    [1 for _ in all_regions],
    index=all_regions
)
```

```python
import numpy as np

your_distribution = df["region"].value_counts()
your_distribution = your_distribution.reindex(all_regions, fill_value=0)

difference = ideal_distribution - your_distribution
difference = difference[difference > 0].sort_values(ascending=False)
print("💡 You could train these regions more:")
print(difference)
```

```python
your_workouts = df["title"].unique().tolist()

suggested = []
for region in difference.index:
    for ex in exerciseLibrary:
        if ex["group"] == region and ex["name"] not in your_workouts:
            suggested.append(ex["name"])
            break

print("\n🔥 Suggested to balance your training:")
for s in suggested:
    print("✅", s)


```
